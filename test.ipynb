```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parámetros de la cúbica base ---
a3_base = 1.0      # coeficiente cúbico nominal
a2 = -0.6
a1 = 0.2
a0 = 0.0

# --- Dominios elegidos para "variaciones efectivas" ---
x = np.linspace(-2.0, 2.0, 600)                 # eje horizontal
da3 = np.linspace(-0.25, 0.25, 300)             # variación leve de a3
A3 = a3_base + da3[:, None]                     # (Ny,1) para broadcasting
X  = x[None, :]                                 # (1,Nx)

print(f"{A3=}")

# --- Cúbica y su variación respecto a la base ---
Y = A3*X**3 + a2*X**2 + a1*X + a0
Y0 = a3_base*X**3 + a2*X**2 + a1*X + a0
D = Y - Y0  # esto aísla el efecto de variar a3  
print(f"{D=}")

# --- Heatmap ---
plt.figure(figsize=(10, 4.8))
im = plt.imshow(
    D,
    origin="lower",
    aspect="auto",
    extent=(x.min(), x.max(), da3.min(), da3.max()),
    interpolation="nearest"
)
plt.colorbar(im, label=r"$\Delta y = y(x,\Delta a_3) - y(x,0)$")
plt.xlabel("x")
plt.ylabel(r"$\Delta a_3$ (variación del coeficiente cúbico)")
plt.title(r"Mapa de calor del efecto de variar levemente $a_3$ en $y=a_3x^3+a_2x^2+a_1x+a_0$")
plt.tight_layout()
plt.show()
```
