```python
import numpy as np
import matplotlib.pyplot as plt

# Crear una malla de puntos en R2
x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
X, Y = np.meshgrid(x, y)

# Calcular la función Z = f(X, Y)
# Z = np.cosh(X) - np.cosh(Y)
Z = X**2 + Y**2

print(f"shape of x: {x.shape}\nshape of y: {y.shape}\n\nshape of X: {X.shape}\nshape of Y: {Y.shape}\nshape of Z: {Z.shape}")
```

```python
# Visualizar la malla de puntos
plt.figure(figsize=(8, 8))
plt.plot(X, Y, 'bo', markersize=4)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Malla de puntos en R²')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
plt.imshow(Z, origin='lower', cmap='hot', aspect='auto', interpolation='bilinear', alpha=0.9)
plt.colorbar(label='f(x, y)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Mapa de calor: f(x,y) = x² + y²')
plt.grid(True, color='black', linewidth=0.5, alpha=0.1)
plt.show()
```

```python
fig = plt.figure(figsize=(16, 12))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='hot', alpha=0.9, edgecolor='none', rstride=1, cstride=1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('f(x, y)')
ax.set_title('Gráfico 3D: f(x,y) = cosh(x) - cosh(y)')
plt.show()
```
