# Polinomial Fitting
---

```python
# 5 random data points
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python

x = np.cumsum(np.random.rand(5) * 10)
y = np.random.rand(5) * 10
plt.scatter(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('5 Random Data Points')
plt.grid()
plt.show()
```

```python
# fit a 3-order polynomial (polyfit returns coefficients)
p = np.polyfit(x, y, 4)
yHat = np.polyval(p, x)  # predicted data is evaluation of polynomial

# create smooth line for polynomial
xx = np.linspace(min(x), max(x), 100)
yy = np.polyval(p, xx) # evaluate polynomial at xx

# plot
plt.scatter(x, y)
plt.scatter(xx, yy, color='red', s=1)
plt.scatter(x, yHat, color='red', marker='x')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Polynomial Fit')
plt.grid()
plt.show()
```

```python
nmax = 10
errors = np.zeros(nmax)
for n in range(1, nmax):
    p = np.polyfit(x, y, n)
    yHat = np.polyval(p, x)  # predicted data is evaluation of polynomial
    
    # compute the error
    errors[n] = np.sum((y - yHat)**2)  # MSE
    
plt.plot(np.arange(0, nmax), errors, "-o")
plt.title("Errors")
plt.xlabel('Poly order')
plt.ylabel('MSE Error')
plt.grid()
plt.show()

# Ignorar Rank Warnings
```
